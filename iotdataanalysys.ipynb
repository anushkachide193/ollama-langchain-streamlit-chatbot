```python
#import lib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("iot_sensor_data.csv")
df["Timestamp"]=pd.to_datetime(df["Timestamp"])
print("IOT sensor Data\n",df)
```

```python
#statical analysis
temp =df["Temperature"].values
mean_temp=np.mean(temp) 
std_temp=np.std(temp)
var_temp=np.var(temp)
max_temp=np.max(temp)
min_temp=np.min(temp)
print("\n Mean Temperature: ",mean_temp)
print("Standard Derivation Temperature:",std_temp)
print("Variance of temperature",var_temp)
print("Maximum temperature",max_temp)
print("Minimum temperature:",min_temp)
```

```python
plt.figure
plt.bar(df["Temperature"],df["Humidity"])
plt.xlabel("Temperature")
plt.ylabel("Humidity")
plt.title("Temperature vs Humidity ")
plt.show()
```
